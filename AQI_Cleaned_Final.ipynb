# Exploratory Data Analysis

```python
# Cleaned Dataset
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\nayan\Downloads\Aqi_Cleaned_final.csv")
```

```python
df.shape
```

```python
# Convert the date column into datetime format
df['date'] = pd.to_datetime(df['date'])
```

Time-Based Analysis

```python
# Count the total number of records for each year
yearly_records = df.groupby(df['date'].dt.year).size()

# Display the result
print(yearly_records)
```

```python
# Calculate the average AQI for each year
yearly_avg_aqi = df.groupby(df['date'].dt.year)['aqi_value'].mean()

# Display the result
print(yearly_avg_aqi)
```

```python
# Calculate the average AQI for each month
monthly_avg_aqi = df.groupby(df['date'].dt.month)['aqi_value'].mean()

# Display the result
print(monthly_avg_aqi)
```

```python
# Create a quarter column from the date
df['quarter'] = df['date'].dt.quarter

# Calculate the average AQI for each quarter
quarterly_avg_aqi = df.groupby('quarter')['aqi_value'].mean()

# Display the result
print(quarterly_avg_aqi)
```

```python
# Count each AQI category for every year
yearly_category_count = df.groupby(
    [df['date'].dt.year, 'calculated_status']
).size()

# Display the result
print(yearly_category_count)
```

```python
# Calculate the average AQI for each month of each year
monthly_aqi_trend = df.groupby(
    [df['date'].dt.year, df['date'].dt.month]
)['aqi_value'].mean()

# Display the result
print(monthly_aqi_trend)
```

State-wise Analysis

```python
# Count the total number of records for each state
state_records = df.groupby('state').size()

# Display the result
print(state_records)
```

```python
# Calculate the average AQI for each state
state_avg_aqi = df.groupby('state')['aqi_value'].mean()

# Display the result
print(state_avg_aqi)
```

```python
# Sort states by average AQI from highest to lowest
top_10_polluted_states = state_avg_aqi.sort_values(ascending=False).head(10)

# Display the top 10 most polluted states
print(top_10_polluted_states)
```

```python
# Sort states by average AQI from lowest to highest
top_10_cleanest_states = state_avg_aqi.sort_values().head(10)

# Display the top 10 cleanest states
print(top_10_cleanest_states)
```

```python
# Count each AQI category for every state
state_category_count = df.groupby(
    ['state', 'calculated_status']
).size()

# Display the result
print(state_category_count)
```

```python
# Find the maximum AQI recorded in each state
state_max_aqi = df.groupby('state')['aqi_value'].max()

# Display the result
print(state_max_aqi)
```

Area-wise Total Records

```python
# Count the total number of records for each area
area_records = df.groupby('area').size()

# Display the result
print(area_records)
```

```python
# Calculate the average AQI for each area
area_avg_aqi = df.groupby('area')['aqi_value'].mean()

# Display the result
print(area_avg_aqi)
```

```python
# Sort areas by average AQI from highest to lowest
top_10_polluted_areas = area_avg_aqi.sort_values(ascending=False).head(10)

# Display the top 10 most polluted areas
print(top_10_polluted_areas)
```

```python
# Sort areas by average AQI from lowest to highest
top_10_cleanest_areas = area_avg_aqi.sort_values().head(10)

# Display the top 10 cleanest areas
print(top_10_cleanest_areas)
```

```python
# Find the maximum AQI recorded in each area
area_max_aqi = df.groupby('area')['aqi_value'].max()

# Display the result
print(area_max_aqi)
```

AQI Category Analysis

```python
# Count the total number of records in each AQI category
category_records = df['calculated_status'].value_counts()

# Display the result
print(category_records)
```

```python
# Calculate the percentage of records in each AQI category
category_percentage = df['calculated_status'].value_counts(normalize=True) * 100

# Display the result
print(category_percentage)
```

```python
# Count Severe AQI records for each state
state_severe = df[df['calculated_status'] == 'Severe'].groupby('state').size()

# Sort from highest to lowest
state_severe = state_severe.sort_values(ascending=False)

# Display the result
print(state_severe)
```

# Summary & Key Findings

## 1. Overall Dataset

- The dataset contains **425,279 air-quality records** from **May 2015 to June 2025**.
- It covers **32 states and 295 areas**.
- The average AQI is **118.66**, while the median AQI is **96**.
- AQI values range from **3 to 500**.

## 2. AQI Category Distribution

- **Satisfactory** is the most common category, accounting for **36.49%** of all records.
- **Moderate** accounts for **32.55%** of records.
- **Good** accounts for **16.43%** of records.
- **Poor** accounts for **10.30%** of records.
- **Very Poor** accounts for **3.62%** of records.
- **Severe** accounts for **0.60%** of records.

## 3. Yearly Trend

- The highest average AQI was recorded in **2016**, at **157.05**.
- Overall AQI showed a **downward trend over the long term**, although it increased in some years.
- The average AQI reached **105.33 in 2024** and **102.76 in 2025**.
- **2025 contains data only up to June 19**, so it should not be directly compared with complete years.

## 4. Seasonal Pattern

- **November** had the highest average AQI at **174.31**.
- **December** also had a high average AQI of **168.17**.
- **August** had the lowest average AQI at **66.16**.
- **Q4** had the highest quarterly average AQI at **154.78**.
- **Q3** had the lowest quarterly average AQI at **67.96**.

## 5. State-wise Findings

- **Delhi** had the highest average AQI at **215.90**.
- **Bihar** had the second-highest average AQI at **165.91**.
- **Uttar Pradesh** had an average AQI of **159.54**.
- **Mizoram** had the lowest average AQI at **45.06**.
- Other states with relatively low average AQI include **Meghalaya, Sikkim, and Puducherry**.

## 6. Area-wise Findings

- **Loni Ghaziabad** had the highest average AQI at **336.50**.
- Other areas with high average AQI include **Jharsuguda, Gurgaon, Delhi, and Ghaziabad**.
- **Tirunelveli** had the lowest average AQI at **31.98**.
- Other areas with low average AQI include **Perundurai, Madikeri, and Palkalaiperur**.

## 7. Poor AQI Records

- **Haryana** had the highest number of Poor AQI records with **9,136**.
- **Uttar Pradesh** had **7,772** Poor AQI records.
- **Bihar** had **5,985** Poor AQI records.
- Rajasthan and Madhya Pradesh also recorded a high number of Poor AQI records.

## 8. Severe AQI Records

- **Uttar Pradesh** had the highest number of Severe AQI records with **1,055**.
- **Bihar** had **573** Severe AQI records.
- **Haryana** had **541** Severe AQI records.
- **Delhi** recorded **163** Severe AQI records.

## Overall Conclusion

The analysis shows that air quality varies significantly across time, states, and areas. Overall AQI showed a downward trend over the long term, but high pollution levels remain concentrated in several regions. Delhi, Bihar, and Uttar Pradesh recorded high average AQI values, while Mizoram and Meghalaya recorded much lower average AQI values. AQI also showed a clear seasonal pattern, with higher values mainly during the last quarter of the year.
